```python
import numpy as np

# Load CSV file, skip header
data = np.genfromtxt(
    'Iris.csv',
    delimiter=',',
    skip_header=1,
    usecols=(1, 2, 3, 4)
)

print("Shape of dataset:", data.shape)
print("Data type of numeric array:", data.dtype)

print("\nFirst 5 rows:")
print(data[:5])
```

```python
columns = [
    "Sepal Length",
    "Sepal Width",
    "Petal Length",
    "Petal Width"
]

for i in range(4):
    mean = np.mean(data[:, i])
    median = np.median(data[:, i])
    std = np.std(data[:, i])

    print(
        f"{columns[i]}: "
        f"Mean={mean:.3f}, "
        f"Median={median:.3f}, "
        f"Std={std:.3f}"
    )
```

```python
filtered = data[
    (data[:, 2] > 1.5) &
    (data[:, 1] < 3.0)
]

print("Filtered rows count:", len(filtered))

print("\nFirst 5 filtered rows:")
print(filtered[:5])
```

```python
minimum = np.min(data, axis=0)
maximum = np.max(data, axis=0)

normalized = (data - minimum) / (maximum - minimum)

print("First 5 rows after normalization:")
print(np.round(normalized[:5], 3))
```

```python
# Create a copy of the data
data_with_missing = data.copy()

# Add some missing values (NaN) for demonstration
data_with_missing[2, 1] = np.nan
data_with_missing[5, 2] = np.nan
data_with_missing[10, 3] = np.nan

# Count missing values
missing_count = np.isnan(data_with_missing).sum()

print("Total missing values:", missing_count)

# Replace missing values with zero
data_with_missing[np.isnan(data_with_missing)] = 0

print("\nData after replacing missing values with zero:")
print(data_with_missing[:15])
```
